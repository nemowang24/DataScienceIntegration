# Converting Jupyter Notebook to HTML
This notebook demonstrates how to convert a Jupyter notebook to HTML format using nbconvert.

```python
!pip install nbcon
```

```python
import os
import nbformat
from nbconvert import HTMLExporter
from nbconvert.preprocessors import ExecutePreprocessor
```

### Define the input and output paths

```python
input_path = r"D:\MyDrive2\pythonprojects\class2\DataMining\homework\final_project\publish_recommendation.ipynb"
output_dir = os.path.dirname(input_path)
output_file = os.path.join(output_dir, "publish_recommendation.html")
```

```python
# Read the notebook
with open(input_path, 'r', encoding='utf-8') as f:
    notebook = nbformat.read(f, as_version=4)

# Create HTML exporter
html_exporter = HTMLExporter()
html_exporter.template_name = 'classic'

# Convert notebook to HTML
body, resources = html_exporter.from_notebook_node(notebook)

# Write the HTML file
with open(output_file, 'w', encoding='utf-8') as f:
    f.write(body)
```

```python
print(f"Notebook has been converted to HTML and saved at: {output_file}")
```

The notebook has been converted to HTML format with all cells, code, and output preserved. You can open the HTML file in any web browser to view the contents.
